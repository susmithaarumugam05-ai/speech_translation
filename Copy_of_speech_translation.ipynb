```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", DEVICE)

MODEL_NAME = "facebook/nllb-200-distilled-600M"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
translator = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME).to(DEVICE)

print("NLLB model loaded successfully")

```

```python
# Whisper → NLLB language mapping
NLLB_LANG_MAP = {
    "en": "eng_Latn",
    "zh": "zho_Hans",
    "ta": "tam_Taml",
    "te": "tel_Telu",
    "ml": "mal_Mlym",
    "hi": "hin_Deva",
    "ko": "kor_Hang",
    "ru": "rus_Cyrl",
    "es": "spa_Latn",
    "de": "deu_Latn"
}

def translate_any_to_any(text, src_whisper_lang, tgt_whisper_lang):
    # No translation needed
    if src_whisper_lang == tgt_whisper_lang:
        return text

    src_lang = NLLB_LANG_MAP.get(src_whisper_lang, "eng_Latn")
    tgt_lang = NLLB_LANG_MAP.get(tgt_whisper_lang, "eng_Latn")

    tokenizer.src_lang = src_lang
    inputs = tokenizer(text, return_tensors="pt").to(DEVICE)

    outputs = translator.generate(
        **inputs,
        forced_bos_token_id=tokenizer.convert_tokens_to_ids(tgt_lang),
        max_length=512
    )

    return tokenizer.batch_decode(outputs, skip_special_tokens=True)[0]
```

```python
import gradio as gr
import subprocess
import pysrt
from gtts import gTTS
import os

# ---------------- LANGUAGE MAP ----------------
UI_LANG_MAP = {
    "English": "en",
    "Tamil": "ta",
    "Hindi": "hi",
    "Malayalam": "ml",
    "Telugu": "te"
}

# gTTS language map (IMPORTANT FIX)
GTTS_LANG_MAP = {
    "en": "en",
    "ta": "ta",
    "hi": "hi",
    "ml": "ml",
    "te": "te"
}

# ---------------- MAIN PIPELINE ----------------
def convert_video(video_input, target_lang_name):
    try:
        # Fix Gradio video path
        if isinstance(video_input, dict):
            video_path = video_input["name"]
        else:
            video_path = video_input

        target_lang = UI_LANG_MAP[target_lang_name]

        audio_path = "audio.wav"
        tts_audio = "tts.mp3"
        subtitle_path = "subs.srt"
        final_video = "final_translated_video.mp4"

        # 1️⃣ Extract audio
        subprocess.run(
            ["ffmpeg", "-i", video_path, "-ac", "1", "-ar", "16000", audio_path, "-y"],
            check=True
        )

        # 2️⃣ Transcribe (Whisper)
        result = whisper_model.transcribe(audio_path)
        detected_lang = result["language"]
        text = result["text"]
        segments = result["segments"]

        # 3️⃣ Translate
        translated_text = translate_any_to_any(
            text,
            detected_lang,
            target_lang
        )

        # 4️⃣ Generate subtitles (clean)
        subs = pysrt.SubRipFile()
        for i, seg in enumerate(segments):
            subs.append(
                pysrt.SubRipItem(
                    index=i + 1,
                    start=pysrt.SubRipTime(seconds=seg["start"]),
                    end=pysrt.SubRipTime(seconds=seg["end"]),
                    text=translated_text
                )
            )
        subs.save(subtitle_path)

        # 5️⃣ Generate translated audio (FIXED)
        tts_lang = GTTS_LANG_MAP.get(target_lang, "en")
        tts = gTTS(translated_text, lang=tts_lang)
        tts.save(tts_audio)

        # 6️⃣ Merge video + audio + styled subtitles
        subtitle_style = (
            "FontSize=28,"
            "Alignment=2,"
            "MarginV=40,"
            "Outline=1,"
            "Shadow=0"
        )

        subprocess.run(
            [
                "ffmpeg",
                "-i", video_path,
                "-i", tts_audio,
                "-vf", f"subtitles={subtitle_path}:force_style='{subtitle_style}'",
                "-map", "0:v",
                "-map", "1:a",
                "-c:v", "libx264",
                "-c:a", "aac",
                "-shortest",
                final_video,
                "-y",
            ],
            check=True
        )

        return (
            f"Detected language: {detected_lang}",
            translated_text,
            final_video
        )

    except Exception as e:
        return (
            "Error occurred",
            str(e),
            None
        )

# ---------------- UI ----------------
ui = gr.Interface(
    fn=convert_video,
    inputs=[
        gr.Video(label="Upload Any Language Video"),
        gr.Dropdown(
            choices=list(UI_LANG_MAP.keys()),
            label="Convert Video To",
            value="English"
        )
    ],
    outputs=[
        gr.Textbox(label="Detected Language"),
        gr.Textbox(label="Translated Text"),
        gr.Video(label="Final Translated Video")
    ],
    title="🎥 AI Multilingual Video Translator",
    description=(
        "Upload a video in any language → "
        "auto-detect speech → translate → "
        "get a new video with translated audio and clean subtitles"
    )
)

ui.launch()
```

```python
import torch
import whisper
import subprocess
import pysrt
from gtts import gTTS
import re

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Device
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", DEVICE)

# Load Whisper
whisper_model = whisper.load_model("small").to(DEVICE)
print("Whisper loaded")

# Load NLLB
MODEL_NAME = "facebook/nllb-200-distilled-600M"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
translator = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME).to(DEVICE)
print("Translator loaded")
```

```python
# UI → internal language codes
UI_LANG_MAP = {
    "English": "en",
    "Tamil": "ta",
    "Hindi": "hi",
    "Malayalam": "ml",
    "Telugu": "te"
}

# gTTS language codes
GTTS_LANG_MAP = {
    "en": "en",
    "ta": "ta",
    "hi": "hi",
    "ml": "ml",
    "te": "te"
}

# NLLB mapping
NLLB_LANG_MAP = {
    "en": "eng_Latn",
    "ta": "tam_Taml",
    "hi": "hin_Deva",
    "ml": "mal_Mlym",
    "te": "tel_Telu",
    "es": "spa_Latn",
    "zh": "zho_Hans",
    "ko": "kor_Hang",
    "ru": "rus_Cyrl",
    "de": "deu_Latn"
}
```

```python
def translate_any_to_any(text, src_lang, tgt_lang):
    src = NLLB_LANG_MAP.get(src_lang, "eng_Latn")
    tgt = NLLB_LANG_MAP.get(tgt_lang, "eng_Latn")

    tokenizer.src_lang = src
    inputs = tokenizer(text, return_tensors="pt").to(DEVICE)

    output = translator.generate(
        **inputs,
        forced_bos_token_id=tokenizer.convert_tokens_to_ids(tgt),
        max_length=512
    )

    return tokenizer.batch_decode(output, skip_special_tokens=True)[0]


def convert_video(video_input, target_lang_name):
    try:
        video_path = video_input["name"] if isinstance(video_input, dict) else video_input
        target_lang = UI_LANG_MAP[target_lang_name]

        audio_path = "audio.wav"
        tts_audio = "tts.mp3"
        subtitle_path = "subs.srt"
        final_video = "final_video.mp4"

        # Extract audio
        subprocess.run(
            ["ffmpeg", "-i", video_path, "-ac", "1", "-ar", "16000", audio_path, "-y"],
            check=True
        )

        # Transcribe
        result = whisper_model.transcribe(audio_path)
        detected_lang = result["language"]
        text = result["text"]
        segments = result["segments"]

        # Translate
        translated_text = translate_any_to_any(text, detected_lang, target_lang)

        # Create subtitles (sentence-based)
        sentences = re.split(r'(?<=[.!?।])\s+', translated_text)
        subs = pysrt.SubRipFile()

        for i, seg in enumerate(segments):
            subs.append(
                pysrt.SubRipItem(
                    index=i + 1,
                    start=pysrt.SubRipTime(seconds=seg["start"]),
                    end=pysrt.SubRipTime(seconds=seg["end"]),
                    text=sentences[i] if i < len(sentences) else ""
                )
            )

        subs.save(subtitle_path)

        # Generate TTS
        gTTS(translated_text, lang=GTTS_LANG_MAP[target_lang]).save(tts_audio)

        # Merge video + audio + subtitles
        subtitle_style = (
            "FontName=Noto Sans,"
            "FontSize=28,"
            "Alignment=2,"
            "MarginV=40,"
            "Outline=1"
        )

        subprocess.run(
            [
                "ffmpeg",
                "-i", video_path,
                "-i", tts_audio,
                "-vf",
                f"subtitles={subtitle_path}:fontsdir=/usr/share/fonts/truetype/noto:"
                f"force_style='{subtitle_style}'",
                "-map", "0:v",
                "-map", "1:a",
                "-shortest",
                final_video,
                "-y"
            ],
            check=True
        )

        return f"AUTO: {detected_lang.upper()}", translated_text, final_video

    except Exception as e:
        return "Error", str(e), None
```

```python
import gradio as gr

with gr.Blocks() as app:

    gr.Markdown(
        """
        # 🎥 AI Multilingual Video Translator
        **Translate with ease**
        Upload a video in any language → auto-detect speech → translate →
        get a new video with translated audio and clean subtitles.
        """
    )

    with gr.Row():
        with gr.Column(scale=1):
            gr.Markdown("### 📹 Video Input")

            video_input = gr.Video(label="Upload Any Language Video")
            target_lang = gr.Dropdown(
                choices=list(UI_LANG_MAP.keys()),
                value="English",
                label="Convert Video To"
            )

            translate_btn = gr.Button("✨ Translate Video", variant="primary")
            reset_btn = gr.Button("Reset")

        with gr.Column(scale=1):
            detected_lang_box = gr.Textbox(label="Detected Language", interactive=False)
            translated_text_box = gr.Textbox(label="Translated Text", lines=8, interactive=False)
            final_video = gr.Video(label="Final Translated Video")

    translate_btn.click(
        fn=convert_video,
        inputs=[video_input, target_lang],
        outputs=[detected_lang_box, translated_text_box, final_video]
    )

    reset_btn.click(
        fn=lambda: (None, "", "", None),
        outputs=[video_input, detected_lang_box, translated_text_box, final_video]
    )

app.launch()
```
